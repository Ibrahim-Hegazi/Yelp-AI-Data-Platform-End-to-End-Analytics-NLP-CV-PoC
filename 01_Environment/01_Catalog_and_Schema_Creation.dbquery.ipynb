```sql
--=====================================================================================
-- CATALOG CREATION
--=====================================================================================

CREATE CATALOG IF NOT EXISTS yelp_poc_uc;

--=====================================================================================
-- SCHEMA CREATION (Bronze, Silver, Gold)
--=====================================================================================

CREATE SCHEMA IF NOT EXISTS yelp_poc_uc.Bronze
COMMENT 'Raw ingested data from Yelp sources with minimal transformation';

CREATE SCHEMA IF NOT EXISTS yelp_poc_uc.Silver
COMMENT 'Cleansed and enriched data with applied business rules and quality checks';

CREATE SCHEMA IF NOT EXISTS yelp_poc_uc.Gold
COMMENT 'Curated analytics-ready tables optimised for reporting and some ML consumption';

--=====================================================================================
-- VOLUME CREATION (Landing Zone for Raw Extracts)
--=====================================================================================

CREATE VOLUME IF NOT EXISTS yelp_poc_uc.bronze.landing_zone
COMMENT 'Volume for storing initial raw data extracted from main Yelp data sources';

CREATE VOLUME IF NOT EXISTS yelp_poc_uc.bronze.validated_schema_zone
COMMENT 'Volume for storing schema-validated raw data ready for Bronze table ingestion';

--=====================================================================================
-- SET DEFAULT CONTEXT
--=====================================================================================

-- USE CATALOG yelp_poc_uc;
-- USE SCHEMA Bronze;
-- USE SCHEMA Silver;
-- USE SCHEMA Gold;
```
